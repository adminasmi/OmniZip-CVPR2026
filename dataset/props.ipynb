#### image sizes of datasets

```python
import os
import time
import subprocess
from pathlib import Path
from PIL import Image
from tqdm import tqdm

import pandas as pd
```

```python
""" 1. CLIC """
dataset_dir = '/media/ps/ssd6/zhaoy/datasets/clic/mobile/valid'
dataset_dir = Path(dataset_dir)

images = list(dataset_dir.glob('*.png'))
rows = []
for image in images:
    with Image.open(image) as I:
        width, height = I.size
        npixels = width * height
        rows.append([image.name, width, height, npixels])
        
df = pd.DataFrame(rows, columns=['image', 'width', 'height', 'npixels'])
display(df)
display(df.describe())
```
